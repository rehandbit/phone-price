```python
def fix_device_type(x):
    x = str(x).lower().replace(" ", "")
    if any(brand in x for brand in ['moto', 'galaxy', 'iphone', 'pixel', 'smart']):
        return 'Smart Phone'
    if 'java' in x:
        return 'Java Phone'
    if 'feature' in x:
        return 'Feature Phone'
    if 'smart' in x:
        return 'Smart Phone'
    if 'Symbian' in x:
        return 'Smart Phone'
    return np.nan


df['device_type'] = df['device_type'].apply(fix_device_type)

df.device_type.unique()
```

creating function to convert string to number and other ram storage into one UNIT and battery power into simple unit

```python
# convert into number
def into_number(column, pattern= r"([\d.]+)"):
    text = column.astype(str).str.replace("," , "", regex=False)
    return pd.to_numeric(text.str.extract(pattern)[0], errors="coerce")

# for ram gb tb pb , converting into gb
def to_gb(col):
    n = into_number(col)
    n = n.where(~col.str.contains("MB", na=False), n/1024)
    n = n.where(~col.str.contains("TB", na=False), n* 1024)
    return n

import re
def max_watt(text):
    if not isinstance(text, str):
        return np.nan
    w = re.findall(r"(\d+(?:\.\d+)?)\s*W", text)
    return max(map(float, w)) if w else np.nan
```
